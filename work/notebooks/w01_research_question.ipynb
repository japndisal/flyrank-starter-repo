## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
import pandas as pd

path = "https://raw.githubusercontent.com/japndisal/flyrank-starter-repo/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(path)
print("Shape:", df.shape)

# Trend breakdown
trend_counts = df["trend_direction"].value_counts(dropna=False)
print("\nTrend direction counts:")
print(trend_counts)

# Simple flags
down = df["trend_direction"] == "down"
high_imp = df["impressions_90d"] >= 500
high_sess = df["sessions_90d"] >= 30
both = down & high_imp

print(f"\nPages trending down:            {down.sum():,} ({down.mean():.1%})")
print(f"Pages with >=500 impressions:   {high_imp.sum():,} ({high_imp.mean():.1%})")
print(f"Pages with >=30 sessions:       {high_sess.sum():,} ({high_sess.mean():.1%})")
print(f"Down AND >=500 impressions:     {both.sum():,} ({both.mean():.1%})")
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

**Quick look at the starter data**

The starter file has 30,000 pages. Three numbers make this lane look worth the next 7 weeks:

1. **54.2% of pages are trending down** (by the simple `trend_direction == "down"` proxy), so there is a large pool of weaker pages that could go into a refresh queue.
2. **16,726 pages (about 55.8%) have at least 500 impressions in 90 days**, and 7,114 (about 23.7%) have at least 30 sessions. Many pages have real demand, so small improvements could matter.
3. **9,961 pages are both trending down and have at least 500 impressions (33.2% of the site).** These are the pages where decline and demand overlap, so they are the first candidates for review.

Together these suggest there are plenty of pages with both demand and signs of weakness, so deciding which to review first is a useful problem. These are observed counts from one dataset, not proof that a refresh would fix anything.
