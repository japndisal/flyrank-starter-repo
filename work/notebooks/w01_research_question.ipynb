This notebook frames the first decision I want the analysis to support. I am starting with evidence from the 30,000-row starter dataset rather than assuming that a model is the answer.

## 1. My lane (or freestyle) and why

**Lane: Ranking Signal Analysis.** I chose this lane because the starter data contains observable search and content signals that can be compared to actual page-level performance. My initial finding is that search volume alone does not appear to explain impressions well in this sample. I want to investigate which observable signals are more useful for prioritising pages for review, while keeping the output interpretable enough for a content or SEO practitioner to act on.

```python
import pandas as pd

DATA_PATH = 'data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(DATA_PATH)
print(f'Starter data: {len(df)} rows × {df.shape[1]} columns')
```

## 2. The question: decision, action, cost of a wrong call

**Decision:** Which pages should receive priority for SEO/content investigation based on the search and performance signals available before the decision?

**Action:** A content or SEO practitioner would use the resulting evidence/ranking to decide which pages to review first, then investigate the page and choose an appropriate optimisation action.

**Cost of a wrong call:** Prioritising a low-value page wastes limited analyst/editor time and can cause a genuinely useful opportunity to be overlooked. The ranking therefore needs to be treated as decision support, not as an automatic instruction to change a page.

**Research question:** Which observable signals are most useful for ranking pages for review, and does combining multiple signals provide more useful prioritisation than relying on a single metric such as search volume?

```python
corr = df[['search_volume', 'impressions_90d']].corr().loc['search_volume', 'impressions_90d']
print(f'Search volume → impressions_90d Pearson correlation: {corr:.3f}')
```

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains **30,000 rows and 44 columns**. In an initial check, the Pearson correlation between `search_volume` and `impressions_90d` is **0.001**. This is essentially no linear association in this sample. That makes a single-metric strategy based on search volume worth questioning, but it does not prove that search volume is irrelevant or that there is no non-linear relationship.

These numbers are useful for the next seven weeks because they motivate a broader signal audit: I need to compare multiple observable signals and evaluate whether they improve prioritisation rather than assuming that the largest search-volume keywords should always be reviewed first.

## 4. Careful words: what I can and can't claim

I can claim **observed** relationships and differences in this dataset, and I can produce **directional, decision-support** evidence about which signals are associated with stronger or weaker page outcomes. I can compare a transparent baseline with more complex approaches if the evidence justifies it.

I cannot claim that a correlation proves causation, that a signal will cause a page to gain traffic, or that the analysis can predict Google's future decisions. In particular, the 0.001 correlation only describes the linear relationship between search volume and `impressions_90d` in this sample. It does not establish that search volume has no effect under every circumstance.

This is not just a **train-a-model** exercise because the model is not the deliverable. The goal is to improve a real prioritisation decision. A simple rule, grouped comparison, ranking, or dashboard may be better than ML if it provides the same decision value more transparently. If a model is used later, it must earn its place by improving an agreed evaluation metric over a clear baseline and by producing recommendations that someone can act on.
